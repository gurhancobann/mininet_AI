```python
df1=pd.read_csv("data21.11.2024-topo1.csv")
df2=pd.read_csv("data21.11.2024-topo2.csv")
df3=pd.read_csv("data21.11.2024.csv")
```

```python
df1=pd.DataFrame(df1)
df2=pd.DataFrame(df2)
df3=pd.DataFrame(df3)
```
